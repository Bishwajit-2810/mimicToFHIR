# CSV Inventory and Preview

This notebook scans every `*.csv.gz` and `*.csv` file in `hosp` and `icu`, shows a compact inventory, and previews the first row of each dataset.

The scan covers both compressed and plain CSV files so nothing is missed.

```python
from pathlib import Path

from IPython.display import Markdown, display

import pandas as pd

base_dir = Path("/home/bk/code/Office/test/mimic-iv-clinical-database-demo-2.2")

for folder_name in ("hosp", "icu"):
    folder = base_dir / folder_name
    csv_paths = sorted({*folder.glob("*.csv.gz"), *folder.glob("*.csv")}, key=lambda path: path.name)

    display(Markdown(f"## {folder_name.upper()} ({len(csv_paths)} CSV files)"))

    summary_rows = []
    for csv_path in csv_paths:
        columns = pd.read_csv(csv_path, nrows=0).columns.tolist()
        preview = ', '.join(columns[:6]) + (' ...' if len(columns) > 6 else '')
        summary_rows.append({
            "file": csv_path.name,
            "columns": len(columns),
            "preview": preview,
        })

    display(pd.DataFrame(summary_rows))

    for csv_path in csv_paths:
        display(Markdown(f"### {csv_path.name}"))
        sample_df = pd.read_csv(csv_path, nrows=1)
        print(sample_df.head(1).T.to_string())
```
